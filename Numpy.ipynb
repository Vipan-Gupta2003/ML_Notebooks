```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],

    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],

    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
np.sum(s,axis=0) # column
```

```python
np.sum(s,axis = 1) # row
```

```python
for i in range(0,9,3):
  for j in range(0,9,3):
    n = s[i:i+3,j:j+3]
    print(n.sum())
```

```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
# shape of the matrix
data.shape
```

```python
# average age of student
np.mean(data[:,0])
```

```python
# Extract Math Marks Of All Students
data[:,1]
```

```python
# Extract High Science Score
np.max(data[:,2])
```

```python
# get detailsof student who schored more than 90 in maths
bool_index = data[:,1]>90
bool_index
```

```python
data[data[:,1]>90]
```

```python
# increase all students mathmark by 5
data[:,1]+=5
```

```python
# Findhow many student younger than 19
data[data[:,0]<19]
```

```python
# Average marks in each subject
np.mean(data[:,1:],axis=0)
```

```python
# get data of student who scored at least 80 in both subjects
data[(data[:,1]>=80) & (data[:,2]>=80)]
```

```python
# Replace all science marks < 75 with 0
data[:,2][data[:,2]<75 ] = 0
data
```
